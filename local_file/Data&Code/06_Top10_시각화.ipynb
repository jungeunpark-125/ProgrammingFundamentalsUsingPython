# [6] Top 10 Risk Areas + Visualization

## Absolute risk vs. Rate-based risk
- **Absolute risk (shortage_50/60)**: actual number of missing ballots. Large urban dongs rank highest even with modest overage, because their base population (registered voters) is large. Relevant for deciding where to stockpile extra ballots.
- **Rate-based risk (election-day turnout rate %)**: actual usage as a % of registered voters. Small rural areas rank highest — absolute numbers are small, but the 50-60% assumption was way off. Relevant for deciding where turnout forecasting is least reliable.

These two metrics point to different places, which is itself an interesting finding (urban = logistics problem, rural = forecasting problem).

```python
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import pandas as pd

fm.fontManager.addfont("/usr/share/fonts/opentype/noto/NotoSerifCJK-Regular.ttc")
plt.rcParams["font.family"] = "Noto Serif CJK KR"
plt.rcParams["axes.unicode_minus"] = False

df = pd.read_csv("05_지방선거_읍면동_통합_6to8회.csv")
df["당일투표율(%)"] = df["당일투표율"] * 100
df["선거"] = df["선거명"].str.extract(r"(제\d+회)")
ELECTION_EN = {"제6회":"6th (2014)", "제7회":"7th (2018)", "제8회":"8th (2022)"}
df.shape
```

```python
summary = df.groupby("선거").agg(
    shortage_dongs_50=("부족_50", lambda s: (s > 0).sum()),
    shortage_dongs_60=("부족_60", lambda s: (s > 0).sum()),
).reindex(["제6회","제7회","제8회"])

fig, ax = plt.subplots(figsize=(7,5))
x = range(len(summary))
w = 0.35
ax.bar([i-w/2 for i in x], summary["shortage_dongs_50"], width=w, label="50% threshold", color="#e15759")
ax.bar([i+w/2 for i in x], summary["shortage_dongs_60"], width=w, label="60% threshold", color="#4e79a7")
ax.set_xticks(list(x)); ax.set_xticklabels([ELECTION_EN[e] for e in summary.index])
ax.set_ylabel("Number of dongs with ballot shortage")
ax.set_title("Ballot Shortage Count by Election and Threshold")
for i,(v50,v60) in enumerate(zip(summary["shortage_dongs_50"], summary["shortage_dongs_60"])):
    ax.text(i-w/2, v50+15, str(int(v50)), ha="center", fontsize=9)
    ax.text(i+w/2, v60+15, str(int(v60)), ha="center", fontsize=9)
ax.legend()
plt.tight_layout()
plt.savefig("../Visualization/A_선거별_부족동_개수.png", dpi=150)
plt.show()
```

```python
data = [df[df["선거"]==e]["당일투표율(%)"] for e in ["제6회","제7회","제8회"]]
fig, ax = plt.subplots(figsize=(7,5))
bp = ax.boxplot(data, tick_labels=[ELECTION_EN[e] for e in ["제6회","제7회","제8회"]], patch_artist=True, showfliers=True)
for patch, color in zip(bp["boxes"], ["#e15759","#f28e2b","#4e79a7"]):
    patch.set_facecolor(color); patch.set_alpha(0.6)
ax.axhline(50, color="red", linestyle="--", linewidth=1, label="50% threshold")
ax.axhline(60, color="darkred", linestyle=":", linewidth=1, label="60% threshold")
ax.set_ylabel("Election-day turnout rate (%)")
ax.set_title("Distribution of Election-Day Turnout Rate by Election\n(Shifts downward from 6th to 8th)")
ax.legend()
plt.tight_layout()
plt.savefig("../Visualization/B_선거별_당일투표율_분포.png", dpi=150)
plt.show()
```

## Top 10 by election (not combined)

Combining all three elections hid the 7th/8th results because the 6th had much higher election-day turnout overall.
Below, each election gets its own Top 10 panel. For the absolute-risk chart, the 8th election had **zero** dongs
that actually exceeded the 50% threshold, so we show the unclipped margin (`election-day votes - registered voters x 50%`)
instead of the clipped shortage — negative bars mean "closest to breaching the threshold, but still under it".

```python
ELECTIONS = ["제6회","제7회","제8회"]
TITLE_EN = {"제6회":"6th (2014)", "제7회":"7th (2018)", "제8회":"8th (2022)"}
df["margin_50"] = df["선거일투표수"] - df["준비_50"]

# English place-name labels per election (matches sort order computed below)
PLACE_EN_ABS = {
    "제6회": ["Jeju | Seogwipo-si | Namwon-eup","Seoul | Yangcheon-gu | Mok 5-dong",
              "Seoul | Songpa-gu | Jamsil 3-dong","Seoul | Yangcheon-gu | Sinjeong 6-dong",
              "Gyeonggi | Suwon Jangan-gu | Jeongja 3-dong","Seoul | Guro-gu | Sindorim-dong",
              "Seoul | Nowon-gu | Wolgye 3-dong","Seoul | Seongbuk-gu | Gireum 1-dong",
              "Seoul | Gwangjin-gu | Gwangjang-dong","Seoul | Guro-gu | Gaebong 2-dong"],
    "제7회": ["Seoul | Songpa-gu | Jamsil 3-dong","Jeju | Seogwipo-si | Namwon-eup",
              "Gangwon | Goseong-gun | Hyeonnae-myeon","Seoul | Yangcheon-gu | Sinjeong 6-dong",
              "Gangwon | Goseong-gun | Geojin-eup","Gyeonggi | Gwacheon-si | Galhyeon-dong",
              "Gangwon | Samcheok-si | Hajang-myeon","Gyeongbuk | Yeongcheon-si | Gogyeong-myeon",
              "Gyeonggi | Seongnam Bundang-gu | Yatap 2-dong","Gyeongnam | Uiryeong-gun | Yongdeok-myeon"],
    "제8회": ["Gangwon | Cheorwon-gun | Geunbuk-myeon","Gyeongnam | Uiryeong-gun | Yugok-myeon",
              "Gyeonggi | Yeoncheon-gun | Jung-myeon","Jeonnam | Yeosu-si | Hwajeong-myeon",
              "Gangwon | Goseong-gun | Hyeonnae-myeon","Incheon | Ganghwa-gun | Seodo-myeon",
              "Gyeonggi | Paju-si | Jangdan-myeon","Gyeonggi | Yeoncheon-gun | Jangnam-myeon",
              "Gyeonggi | Yeoncheon-gun | Wangjing-myeon","Gyeongnam | Uiryeong-gun | Nakseo-myeon"],
}
PLACE_EN_RATE = {
    "제6회": ["Gyeongbuk | Cheongsong-gun | Budong-myeon","Gyeongbuk | Yeongyang-gun | Cheonggi-myeon",
              "Gyeongnam | Goseong-gun | Gaecheon-myeon","Jeonnam | Goheung-gun | Pungyang-myeon",
              "Gyeongbuk | Yecheon-gun | Gamcheon-myeon","Chungbuk | Boeun-gun | Sanoe-myeon",
              "Gyeongbuk | Bonghwa-gun | Bongseong-myeon","Gangwon | Goseong-gun | Hyeonnae-myeon",
              "Gyeongbuk | Gunwi-gun | Sobo-myeon","Gyeonggi | Yeoncheon-gun | Wangjing-myeon"],
    "제7회": ["Gangwon | Goseong-gun | Hyeonnae-myeon","Gangwon | Samcheok-si | Hajang-myeon",
              "Gyeongnam | Uiryeong-gun | Yugok-myeon","Gangwon | Cheorwon-gun | Geunbuk-myeon",
              "Gyeongnam | Goseong-gun | Gaecheon-myeon","Gyeongnam | Uiryeong-gun | Yongdeok-myeon",
              "Gyeonggi | Paju-si | Gunnae-myeon","Gyeongbuk | Goryeong-gun | Deokgok-myeon",
              "Gyeonggi | Yeoncheon-gun | Wangjing-myeon","Gyeongbuk | Cheongsong-gun | Pacheon-myeon"],
    "제8회": ["Jeonnam | Yeosu-si | Hwajeong-myeon","Gangwon | Goseong-gun | Hyeonnae-myeon",
              "Gyeongnam | Uiryeong-gun | Yugok-myeon","Gyeongnam | Hapcheon-gun | Ssangchaek-myeon",
              "Seoul | Gangnam-gu | Dogok 2-dong","Gyeonggi | Yeoncheon-gun | Wangjing-myeon",
              "Gangwon | Pyeongchang-gun | Bangnim-myeon","Gyeonggi | Seongnam Bundang-gu | Unjung-dong",
              "Gyeonggi | Seongnam Bundang-gu | Imae 1-dong","Chungnam | Geumsan-gun | Gunbuk-myeon"],
}
```

```python
fig, axes = plt.subplots(1, 3, figsize=(20,6.5))
for ax, e in zip(axes, ELECTIONS):
    sub = df[df["선거"]==e].sort_values("margin_50", ascending=False).head(10).iloc[::-1]
    labels = list(reversed(PLACE_EN_ABS[e]))
    colors = ["#e15759" if v>0 else "#4e79a7" for v in sub["margin_50"]]
    bars = ax.barh(labels, sub["margin_50"], color=colors)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_title(TITLE_EN[e])
    xmin, xmax = sub["margin_50"].min(), sub["margin_50"].max()
    pad = max(abs(xmin), abs(xmax)) * 0.35 + 20
    ax.set_xlim(xmin - pad*0.3, xmax + pad)
    for b, v in zip(bars, sub["margin_50"]):
        ax.text(v + (pad*0.15 if v>=0 else -pad*0.15), b.get_y()+b.get_height()/2, f"{v:,.0f}",
                va="center", ha="left" if v>=0 else "right", fontsize=8)
fig.suptitle("Top 10 Areas by Absolute Shortage Risk, by Election", fontsize=14)
fig.text(0.5, 0.02, "Ballots = Election-day votes - Registered voters x 50%   (positive = shortage, negative = buffer)",
          ha="center", fontsize=10)
plt.tight_layout(rect=[0,0.06,1,0.95])
plt.savefig("../Visualization/C_Top10_절대량위험.png", dpi=150, bbox_inches="tight")
plt.show()
```
